7) Take 3 int inputs from user and check and print the result.
	all are equal
	any of two are equal
	( use and or )

```python
inp1 = input("Enter a number: ")
inp2 = input("Enter another number: ")
inp3 = input("Enter a third number: ")

if inp1 == inp2 == inp3:
    print("All numbers are the same.")
elif inp1 == inp2 or inp1 == inp3 or inp2 == inp3:
    print("Two numbers are the same.")
else:
    print("All numbers are different.")
```

8) Accept two numbers and print the greatest between them

```python
inp1 = int(input("Enter a number: "))
inp2 = int(input("Enter another number: "))

if inp1 > inp2:
    print(f"{inp1} is greater than {inp2}.")
elif inp1 < inp2:
    print(f"{inp1} is less than {inp2}.")
else:
    print(f"{inp1} is equal to {inp2}.")
```

11) Accept an integer and check whether it is an even number or odd.

```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print(f"{num} is an even number.")
else:
    print(f"{num} is an odd number.")
```

12) Accept name and age from the user. Check if the user is a valid voter or not.
      Vaid - Hello Shery, You are a valid voter.
      Invalid - Sorry Shery, you can't cast the vote.
	Part 2 - Print after how many years the user will be eligible

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))

if age >= 18:
    print(f"hello {name}, You are a valid voter.")
else:
    print(f"hello {name}, You are not a valid voter. You will be eligible to vote in {18 - age} years.")
```

14) Accept a year and check if it a leap year or not (google to find out what's a leap year)

```python
year = int(input("Enter a year: "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```
