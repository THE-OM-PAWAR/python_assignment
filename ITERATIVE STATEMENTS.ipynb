**23) Print all the numbers which are either divisible by 3 or 5 in a range**

```python
intj = int(input("Enter an integer: "))

for i in range(1, intj + 1):
    if i % 3 == 0:
        print(f"{i} is divisible by 3")
    elif i % 5 == 0:
        print(f"{i} is divisible by 5")
```
