### 2206. Divide Array Into Equal Pairs

```python
nums = [3,2,3,2,2,2,3]

ans = True

for i in nums:
    if nums.count(i) % 2 != 0:
        ans = False
        break

print(ans)
```

```python
from collections import Counter

nums = [3,2,3,2,2,2]

count = Counter(nums)
print(count.values())

print(all(v % 2 == 0 for v in count.values()))
```

### 2341. Maximum Number of Pairs in Array

```python
from collections import Counter

nums = [1,3,2,1,3,2,2,8]

count = Counter(nums)

pairs = sum(v // 2 for v in count.values())

print([pairs, len(nums) // 2 - pairs])
```

```python
nums = [1,3,2,1,3,2,2]

freq = {}

for i in nums:
    freq[i] = freq.get(i, 0) + 1

pairs = 0

print(freq.values())
for value in freq.values():
    pairs += value // 2

leftover = len(nums) - (pairs * 2)

print([pairs, leftover]) 
```

### 2357. Make Array Zero by Subtracting Equal Amounts

```python
nums = [1,5,0,3,5]

print(len(set(i for i in nums if i > 0)))
```

### 2293. Min Max Game - Array

```python
nums = [1,3,5,2,4,8,2,2]

while len(nums) > 1:

    new = []
    for i in range(len(nums) // 2): 
        if i % 2 == 0:
            new.append(min(nums[2*i], nums[2*i + 1]))
        else:
            new.append(max(nums[2*i], nums[2*i + 1]))

    nums = new

print(nums[0])
```

### 2164. Sort Even and Odd Indices Independently

```python
nums = [2,4,6,1,1,6,5,1,8,2]

arr = [nums[0]]

for i in range(1, len(nums)):
    if nums[i] != nums[i - 1]:
        arr.append(nums[i])

count = 0

for i in range(1, len(arr)-1) : 

    if arr[i] > arr[i-1] and arr[i] > arr[i+1] :
        count += 1
    elif arr[i] < arr[i-1] and arr[i] < arr[i+1] : 
        count += 1

print(count)

```
